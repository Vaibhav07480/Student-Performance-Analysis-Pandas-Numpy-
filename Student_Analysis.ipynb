```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Students_Performance.csv")

df.head()
```

```python
df.isnull().sum()
df.info()
df.shape
df.describe()
```

```python
# Basic analysis
df.columns
df["Gender"].value_counts()
df["Age"].value_counts().sort_index()

df["Attendance (%)"].mean()
```

```python
#Top students
top_students = df.sort_values("Average_Grade", ascending=False)

top_students[["Student_Name", "Average_Grade"]].head(10)
```

```python
# student with good attenden
good_attendance = df[df["Attendance"] >= 90]

good_attendance[["Student_Name", "Attendance", "Average_Grade"]]
```

```python
plt.scatter(df["Study_Hours"], df["Average_Grade"])

plt.xlabel("Study Hours")
plt.ylabel("Average Grade")
plt.title("Study Hours vs Average Grade")

plt.show()
```

```python
plt.scatter(df["Attendance (%)"], df["Grade"])

plt.xlabel("Attendance")
plt.ylabel("Average Grade")
plt.title("Attendance vs Average Grade")

plt.show()
```

```python
#top performing students
top_students = df.sort_values("Total_Score", ascending=False)

top_students[["First_Name", "Last_Name", "Total_Score", "Grade"]].head(10)
```

```python
department_scores = df.groupby("Department")["Total_Score"].mean().sort_values(ascending=False)

department_scores
```

```python
sample_df = df.sample(n=500, random_state=42)
plt.scatter(sample_df["Attendance (%)"], sample_df["Final_Score"])

plt.title("Attendance vs Final Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Final Score")
plt.show()
```

```python
attendance_by_grade = df.groupby("Grade")["Attendance (%)"].mean()

print(attendance_by_grade)
```

```python
attendance_by_grade.plot(kind="bar", color="green")

plt.title("Average Attendance by Grade")
plt.xlabel("Grade")
plt.ylabel("Attendance (%)")
plt.xticks(rotation=0)

plt.show()
```

```python
# Students with low performance
low_performers = df[df["Total_Score"] < 50]

print(low_performers[[
    "First_Name",
    "Last_Name",
    "Total_Score",
    "Grade"
]].head(10))
```

```python
scores = df["Total_Score"].to_numpy()

print("Maximum Score:", np.max(scores))
print("Minimum Score:", np.min(scores))
print("Average Score:", np.mean(scores))
print("Standard Deviation:", np.std(scores))
```
